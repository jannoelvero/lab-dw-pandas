### Data Description

- Customer - Customer ID

- ST - State where customers live

- Gender - Gender of the customer

- Education - Background education of customers 

- Customer Lifetime Value - Customer lifetime value(CLV) is the total revenue the client will derive from their entire relationship with a customer. In other words, is the predicted or calculated value of a customer over their entire duration as a policyholder with the insurance company. It is an estimation of the net profit that the insurance company expects to generate from a customer throughout their relationship with the company. Customer Lifetime Value takes into account factors such as the duration of the customer's policy, premium payments, claim history, renewal likelihood, and potential additional services or products the customer may purchase. It helps insurers assess the long-term profitability and value associated with retaining a particular customer.

- Income - Customers income

- Monthly Premium Auto - Amount of money the customer pays on a monthly basis as a premium for their auto insurance coverage. It represents the recurring cost that the insured person must pay to maintain their insurance policy and receive coverage for potential damages, accidents, or other covered events related to their vehicle.

- Number of Open Complaints - Number of complaints the customer opened

- Policy Type - There are three type of policies in car insurance (Corporate Auto, Personal Auto, and Special Auto)

- Vehicle Class - Type of vehicle classes that customers have Two-Door Car, Four-Door Car SUV, Luxury SUV, Sports Car, and Luxury Car

- Total Claim Amount - the sum of all claims made by the customer. It represents the total monetary value of all approved claims for incidents such as accidents, theft, vandalism, or other covered events.

```python
import pandas as pd
```

```python
df = pd.read_csv("file1.csv")

df.head()
```

```python
print("Original number of rows:", df.shape[0])
print("Original number of columns:", df.shape[1])

print("\nOriginal column names:")
print(df.columns.tolist())
```

```python
The original dataset contains 4,008 rows and 11 columns. The columns include customer identification, location, demographic information, policy information, income, premiums, complaints, and claim amounts.
```

```python
completely_empty_rows = df.isna().all(axis=1).sum()

print("Completely empty rows:", completely_empty_rows)
```

```python
The dataset contains 2,937 completely empty rows. These rows do not contain any customer information.
```

```python
df = df.dropna(how="all").reset_index(drop=True)

print("Rows after removing empty rows:", df.shape[0])
print("Columns after removing empty rows:", df.shape[1])
```

```python
After removing rows where every value is missing, the dataset contains 1,071 valid customer records and 11 columns. Only completely empty rows were removed.
```

```python
df.columns = (
    df.columns
    .str.strip()
    .str.lower()
    .str.replace(" ", "_", regex=False)
)

print(df.columns.tolist())
```

```python
The column names were standardized by converting them to lowercase and replacing spaces with underscores. This makes the column names easier and more consistent to use in Python.
```

```python
df.dtypes
```

```python
The dataset contains both categorical and numerical variables. Customer, state, gender, education, policy type, and vehicle class are stored as text. Income, monthly premium auto, and total claim amount are numerical. Customer lifetime value and number of open complaints require further attention because they are stored as text.
```

```python
data_type_assessment = pd.DataFrame({
    "current_data_type": df.dtypes.astype(str),
    "variable_type": [
        "Identifier",
        "Categorical",
        "Categorical",
        "Categorical",
        "Numerical",
        "Numerical",
        "Numerical",
        "Coded variable",
        "Categorical",
        "Categorical",
        "Numerical"
    ],
    "action_needed": [
        "No conversion needed",
        "No conversion needed",
        "No conversion needed",
        "No conversion needed",
        "Convert from text to numeric",
        "No conversion needed",
        "No conversion needed",
        "Requires cleaning and interpretation",
        "No conversion needed",
        "No conversion needed",
        "No conversion needed"
    ]
})

data_type_assessment
```

```python
Customer lifetime value has an incorrect data type because it contains percent signs and is stored as text. It should be converted to a numerical column. Number of open complaints also contains slash-separated values such as 1/0/00, so it should not be converted directly until the meaning of each part is confirmed.
```

```python
df["customer_lifetime_value"] = (
    df["customer_lifetime_value"]
    .astype("string")
    .str.replace("%", "", regex=False)
)

df["customer_lifetime_value"] = pd.to_numeric(
    df["customer_lifetime_value"],
    errors="coerce"
)
```

```python
print(
    "Customer lifetime value data type:",
    df["customer_lifetime_value"].dtype
)

print(
    "Missing customer lifetime values:",
    df["customer_lifetime_value"].isna().sum()
)

df["customer_lifetime_value"].head()
```

```python
Customer lifetime value was successfully converted from text to a numerical float64 column. The percent signs were removed, and missing or invalid values were converted to NaN.
```

### Exercise 3

The sales team wants to know if customers with Personal Auto have a lower income than those with Corporate Auto. How does the average income compare between the two policy types?

- Use *loc* to create two dataframes: one containing only Personal Auto policies and one containing only Corporate Auto policies.
- Calculate the average income for each policy.
- Print the results.

```python
# Create a DataFrame for Personal Auto customers
personal_auto = df.loc[df["Policy Type"] == "Personal Auto"]

# Create a DataFrame for Corporate Auto customers
corporate_auto = df.loc[df["Policy Type"] == "Corporate Auto"]

# Calculate the average income
personal_avg_income = personal_auto["Income"].mean()
corporate_avg_income = corporate_auto["Income"].mean()

# Print the results
print("Average Income for Personal Auto:", personal_avg_income)
print("Average Income for Corporate Auto:", corporate_avg_income)
```

```python
if personal_avg_income < corporate_avg_income:
    print("Customers with Personal Auto have a lower average income than customers with Corporate Auto.")
elif personal_avg_income > corporate_avg_income:
    print("Customers with Personal Auto have a higher average income than customers with Corporate Auto.")
else:
    print("Both policy types have the same average income.")
```

When we talk about the top 25%, we are referring to the values that fall above the 75th percentile, which represent the top quarter of the data.
